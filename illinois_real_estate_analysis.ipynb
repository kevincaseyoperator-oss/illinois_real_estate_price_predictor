```python
import pandas as pd

df = pd.read_csv("cleaned_df.csv")
df_illinois = df[df.State == "IL"]
df_illinois = df_illinois.reset_index()
df_illinois = df_illinois.drop(["MarketEstimate"], axis = 1)
df_illinois = df_illinois.drop(["RentEstimate"], axis = 1)
df_illinois = df_illinois.drop(["Latitude"], axis = 1)
df_illinois = df_illinois.drop(["Longitude"], axis = 1)
df_illinois = df_illinois.drop(["index"], axis = 1)
df_illinois

```

# Section 5: Data Science

As stated in Section 4, I want to understand more about how certain traits/ variables of a property affect its listed price. I can use machine learning to make a linear regression model that can take in variables of a property like Bedroom # or Area of property and give a prediction of what that property's listed price will be. In this case, to make a linear regression model I will need to use numeric data that has some correlation to the listed price of a property. Below I also made a correlation table to get a sense of what numeric data would be most useful. Unfortunately the city variable which is a big indicator of listed price isn't numeric, and not every town in Illinois is covered in this data, so the model wouldn't work well for a lot of predictions if that was implemented. PPSq has a strong correlation with listed price, but the listed price is used to calculate that so it wouldn't make sense to use it to predict listed price.

```python
df_illinois.corr(numeric_only=True)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
ind = ["Bedroom", "Bathroom", "Area"]
dep = "ListedPrice"
df_illinois_clean = df_illinois.dropna()
model.fit(df_illinois_clean[ind], df_illinois_clean[dep])
```

```python
myHouse = []
myHouse.append({"Bedroom" : 4, "Bathroom" : 2.5, "Area" : 2600})
df_myHouse = pd.DataFrame(myHouse)
model.predict(df_myHouse)
```

Above I used the model to predict the listed price of my house which was bought in 2006 for $345,000, which is pretty close to the prediction of $353,567. However, adjusted for inflation it should be closer to $550,000 so the model didn't closely predict what the listed price of my house should have been. I think the reason for this discrepancy is the fact that the model didn't take location into account which can cause houses to have very differen t prices even if they are the same size and have the same room count.
